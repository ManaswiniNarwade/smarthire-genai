# 01. Embeddings & Semantic Similarity Exploration
**SmartHire GenAI Capstone Project**

This notebook demonstrates:
1. Loading embedding models (`sentence-transformers` / `all-MiniLM-L6-v2`)
2. Generating dense vector representations of job descriptions and candidate skills
3. Computing cosine similarity and distance metrics
4. Benchmarking semantic matching vs keyword matching

```python
import numpy as np
from sentence_transformers import SentenceTransformer
from sklearn.metrics.pairwise import cosine_similarity

# Load local lightweight embedding model
model = SentenceTransformer('all-MiniLM-L6-v2')
print(f"Embedding Model Dimension: {model.get_sentence_embedding_dimension()}")
```

### Compare Candidate Query with Different Job Descriptions

```python
candidate_profile = "Experienced in Python, LangChain, FAISS, PyTorch, and fine-tuning LLMs for retrieval augmented generation."

job_postings = [
    "Generative AI Engineer: Building LLM pipelines, prompt engineering, FAISS vector search, and RAG architectures in Python.",
    "Frontend Developer: Crafting responsive web apps using React, Next.js, TypeScript, and TailwindCSS.",
    "Data Scientist: Machine learning modeling, statistical analysis, hypothesis testing, and SQL dashboarding.",
    "DevOps Engineer: AWS infrastructure, Kubernetes orchestration, Docker containerization, and Terraform CI/CD."
]

# Compute embeddings
profile_emb = model.encode([candidate_profile], normalize_embeddings=True)
job_embs = model.encode(job_postings, normalize_embeddings=True)

# Compute similarity scores
scores = cosine_similarity(profile_emb, job_embs)[0]

for score, job in sorted(zip(scores, job_postings), key=lambda x: x[0], reverse=True):
    print(f"Similarity: {score * 100:.2f}% | Job: {job[:80]}...")
```
